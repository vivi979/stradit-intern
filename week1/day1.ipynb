# Even and odd

```python
n = int(input("Enter the number: "))
if n%2 == 0:
    print("Even")
else:
    print("odd")
```

```python
def evenodd(n):
    if n%2 == 0:
        return "even"
    else:
        return "odd"
evenodd(3)
```

```python
lst = ["even" if x%2 == 0 else "odd" for x in range(0,11)]
lst
```

# Largest among three numbers

```python
a = 1
b = 4
c = 2

if a>b and a>c:
    print("largest number: ",a)
elif b>c and b>a:
    print("largest number: ",b)
else:
    print("largest number: ",c)
```

```python
def maxnum(a,b,c):
    return max(a,b,c)
maxnum(1,4,2)
```

# factorial of number

```python
def fact(n):
    f = 1
    for i in range(1,n+1):
        f = f*i
    print("Factorial",f)
```

```python
fact(5)
```

# Fibonacci series

```python
def fobi(n):
    a = 0
    b = 1
    print(a)
    print(b)
    for _ in range(n):
        c = a+b
        print(c)
        a = b
        b = c
fobi(10)
```

# Reverse String

```python
st = "string"
st1 = list(st)
for i in range(len(st1)-1,-1,-1):
    print(st1[i], end = "")
```

# Count vowels and consonants

```python
st = input("Enter string")
vowels = "aeiouAEIOU"
vowel, consonents = 0, 0
for ch in st:
    if ch in vowels:
        vowel +=1
    else:
        consonents +=1
print(vowel)
print(consonents)
```

# palindrome string

```python
st = "madam"
class rev:
    def reversee(self,st):
        st1 = ""
        for i in st:
            st1 = i + st1
        return st1
ob = rev()
revs = ob.reversee(st)

if st == revs:
    print("palindrome")
else:
    print("not palindrome")
```
