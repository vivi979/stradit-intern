# 2. array operation

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])

print("Addition",a+b)
print("subtraction",a-b)
print("Multiplication",a*b)
print("Division",a//b)
```

# 3. Reshape Array

```python
arr = np.arange(1,17)
arr1 = arr.reshape(4,4)
print('4x4',arr1)
arr2 = arr.reshape(2,8)
print('2x8',arr2)
```

# 4.statistical analysis

```python
stud_marks = np.array([60,70,80,88,90,93,95,75,78])
print('Mean',np.mean(stud_marks))
print('Median',np.median(stud_marks))
print('Standard deviation',np.std(stud_marks))
print('Maximum marks',np.max(stud_marks))
print('Minimum marks',np.min(stud_marks))
```

# 5. Indexing and Slicing

```python
arr = np.arange(1,17)
arr1 = arr.reshape(4,4)
print(arr1)
print("first row",arr1[0,:])
print("last column",arr1[:,-1])
print("middle 2x2",arr1[1:3,1:3])
```

# 6.Even Numbers

```python
arr = np.arange(1,101)
arr1 = []
for i in range(len(arr)):
    if (arr[i]%2 == 0):
        arr1.append(int(arr[i]))
print('original array',arr)
print('even array',arr1)
```

# 7.Matrix operation

```python
arr = np.arange(1,10)
arr = arr.reshape(3,3)
print(arr)
print('matrix transpose',arr.T)
print('matrix multiplication', np.dot(arr,arr))
print('matrix determinant',np.linalg.det(arr))
print('matrix Inverse',np.linalg.inv(arr))
```

# 8. Missing Value Handling

```python
arr = np.array([78,np.nan,90,66,89,np.nan])
mean_value = np.nanmean(arr)
arr[np.isnan(arr)] = mean_value
arr
```

# 16. Merge DataFrames

```python
employee_df = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Amit", "Sara", "John"]
})

department_df = pd.DataFrame({
    "id": [1, 2, 3],
    "department": ["IT", "HR", "Finance"]
})

merged_df = pd.merge(employee_df, department_df, on="id")

print(merged_df)
```

# 17. Time series Basics

# 23. Sales Dashboard

```python
import pandas as pd
import matplotlib.pyplot as plt

sales_data = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Sales": [1000, 1500, 1200, 1800],
    "Region": ["North", "South", "East", "West"],
    "Product": ["A", "B", "A", "C"]
})

# Monthly Sales Trend
plt.plot(sales_data["Month"], sales_data["Sales"])
plt.title("Monthly Sales Trend")
plt.show()
```

```python
# Top-selling Products
sales_data["Product"].value_counts().plot(kind='bar')
plt.title("Top Selling Products")
plt.show()
```

```python
# Region-wise Sales
sales_data.groupby("Region")["Sales"].sum().plot(kind='bar')
plt.title("Region-wise Sales")
plt.show()
```

# 25. Employee Analytics Dashboard

```python
df = pd.DataFrame({
    "Department": ["IT", "HR", "Finance", "IT", "HR"],
    "Salary": [60000, 50000, 70000, 65000, 52000],
    "Experience": [2, 4, 5, 3, 6]
})
df
```

```python
# Salary Distribution
plt.hist(df["Salary"])
plt.title("Salary Distribution")
plt.show()
```

```python
# Department-wise Average Salary
df.groupby("Department")["Salary"].mean().plot(kind='bar')
plt.title("Average Salary by Department")
plt.show()
```

```python
# Employee Count by Department
df["Department"].value_counts().plot(kind='bar')
plt.title("Employee Count by Department")
plt.show()
```

```python
# Experience vs Salary
plt.scatter(df["Experience"], df["Salary"])
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")
plt.show()
```
