```python
# Student Marks Bar Chart
import matplotlib.pyplot as plt

students = ['Rahul', 'Priya', 'Amit', 'Sara']
marks = [85, 92, 78, 88]

plt.bar(students, marks)

plt.title('Student Marks in Python')
plt.xlabel('Students')
plt.ylabel('Marks')

for i, mark in enumerate(marks):
    plt.text(i, mark + 1, str(mark), ha='center')

plt.show()
```

```python
# Monthly Sales Line Chart
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [10000, 15000, 12000, 18000, 25000]

plt.plot(months, sales, marker='o', label='Sales')

plt.title('Monthly Sales Trend')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.grid(True)
plt.legend()

plt.show()
```

```python
# Product Share Pie Chart
import matplotlib.pyplot as plt

products = ['Mobile', 'Laptop', 'Tablet', 'Watch']
sales = [45, 30, 15, 10]

explode = [0.1, 0, 0, 0]

plt.pie(
    sales,
    labels=products,
    autopct='%1.1f%%',
    explode=explode
)

plt.title('Product Market Share')

plt.show()
```

```python
# Temperature Trend Visualization
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temp = [30, 32, 35, 28, 27, 33, 31]

max_temp = max(temp)
min_temp = min(temp)

max_day = days[temp.index(max_temp)]
min_day = days[temp.index(min_temp)]

plt.plot(days, temp, marker='o')

plt.scatter(max_day, max_temp, s=100, label='Maximum Temp')
plt.scatter(min_day, min_temp, s=100, label='Minimum Temp')

plt.title('Weekly Temperature Trend')
plt.xlabel('Days')
plt.ylabel('Temperature')

plt.legend()

plt.show()
```

```python
# Employee Salary Histogram
import matplotlib.pyplot as plt

salaries = [
    25000, 30000, 35000, 40000,
    45000, 50000, 55000, 60000,
    65000, 70000, 75000, 80000
]

plt.hist(salaries, bins=5)

plt.title('Employee Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Frequency')

plt.show()
```

```python
# Department Salary Comparison
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'department': ['IT','IT','IT','HR','HR','HR',
                   'Finance','Finance','Finance'],
    'salary': [70000,75000,72000,
               50000,52000,51000,
               80000,85000,82000]
}

df = pd.DataFrame(data)

df.boxplot(column='salary', by='department')

plt.title('Department Salary Comparison')
plt.suptitle('')
plt.xlabel('Department')
plt.ylabel('Salary')

plt.show()
```
