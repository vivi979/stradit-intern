```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/ipl.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df['winner'] = df['winner'].fillna('No Result')
df['city'] = df['city'].fillna('Unknown')

df['result_margin'] = df['result_margin'].fillna(df['result_margin'].median())
df['target_runs'] = df['target_runs'].fillna(df['target_runs'].median())
df['target_overs'] = df['target_overs'].fillna(df['target_overs'].median())
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df['toss_match_win'] = (df['toss_winner'] == df['winner'])
```

```python
toss_win_probability = df['toss_match_win'].mean()

print("Toss winner win probability:", toss_win_probability)
```

```python
# H₀ (Null Hypothesis): Toss winners do not have a higher match win probability than 50%
# H₁ (Alternative Hypothesis): Toss winners have a higher match win probability than 50%.
```

```python
# One-Proportion Z-Test
from statsmodels.stats.proportion import proportions_ztest

successes = df['toss_match_win'].sum()
total_matches = len(df)

z_stat, p_value = proportions_ztest(
    count=successes,
    nobs=total_matches,
    value=0.5,
    alternative='larger'
)

print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Toss winners have a higher match win probability than 50%.")
else:
    print("Toss winners do not have a higher match win probability than 50%")
```
